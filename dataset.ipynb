```python
import polars as pl
import xmltodict

with open("./datasets/personlista.xml", encoding="utf-8") as f:
    data = xmltodict.parse(f.read())

persons = data["personlista"]["person"]

records = []
for person in persons:
    records.append({
        "id": person.get("intressent_id", ""),
        "source_id": person.get("sourceid", ""),
        "first_name": person.get("tilltalsnamn", ""),
        "last_name": person.get("efternamn", ""),
        "sort_name": person.get("sorteringsnamn", ""),
        "birth_year": person.get("fodd_ar", ""),
        "gender": person.get("kon", ""),
        "hometown": person.get("iort", ""),
        "party": person.get("parti", ""),
        "constituency": person.get("valkrets", ""),
        "status": person.get("status", ""),
        "url_xml": person.get("person_url_xml", ""),
        "img_80": person.get("bild_url_80", ""),
        "img_192": person.get("bild_url_192", ""),
        "img_max": person.get("bild_url_max", ""),
    })

df = pl.DataFrame(records)
df
```
