```python
import duckdb
import glob
import json
import re
import pandas as pd

def read_cleaned_json(pattern):
    """Read JSON files matching pattern, stripping /* */ block comments."""
    records = []
    for path in sorted(glob.glob(pattern)):
        with open(path, encoding='utf-8') as f:
            content = re.sub(r'/\*.*?\*/', '', f.read(), flags=re.DOTALL)
        try:
            records.append(json.loads(content))
        except json.JSONDecodeError as e:
            print(f"Skipped {path}: {e}")
    return records

con = duckdb.connect()

betankande_df    = pd.json_normalize(read_cleaned_json('datasets/betänkande/*.json'))
motioner_df      = pd.json_normalize(read_cleaned_json('datasets/motioner/*.json'))
propositioner_df = pd.json_normalize(read_cleaned_json('datasets/propositioner/*.json'))

con.register('betankande',    betankande_df)
con.register('motioner',      motioner_df)
con.register('propositioner', propositioner_df)

# See what's actually in each document type
print("=== betänkande ===")
con.sql("SELECT * FROM betankande LIMIT 3").show()
print("=== motioner ===")
con.sql("SELECT * FROM motioner LIMIT 3").show()
print("=== propositioner ===")
con.sql("SELECT * FROM propositioner LIMIT 3").show()
```
